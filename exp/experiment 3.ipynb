```python
import numpy as np

marks = np.array([78, 65, 89, 92, 56, 74, 81, 69, 95, 60])
positions = np.array([1, 3, 5, 8])

selected = marks[positions]

print("Selected Marks:", selected)
print("Average Mark:", np.mean(selected))
```

```python
import numpy as np

salary = np.array([45000, 52000, 60000, 48000, 75000,
                   51000, 39000, 68000, 55000, 47000])

mask = salary > 50000

print("Boolean Mask:", mask)
print("Salaries above ₹50,000:", salary[mask])
print("Number of Employees:", np.sum(mask))
```

```python
import numpy as np

prices = np.array([1500, 2500, 1800, 3200, 1200,
                   2100, 1750, 2800, 900, 1950])

filter_array = prices < 2000
filtered = prices[filter_array]

print("Filtered Prices:", filtered)
print("Minimum Price:", np.min(filtered))
print("Maximum Price:", np.max(filtered))
```

```python
import numpy as np

marks = np.array([45, 78, 92, 34, 67, 88, 56, 95, 41, 73])

selected = marks[[1, 3, 5, 7]]
mask = marks > 75
filtered = marks[(marks >= 50) & (marks <= 80)]

print("Fancy Indexing:", selected)
print("Above 75:", marks[mask])
print("Number above 75:", np.sum(mask))
print("Between 50 and 80:", filtered)
print("Average:", np.mean(filtered))
```

```python
import numpy as np

ages = np.array([12, 25, 67, 45, 72, 34, 18, 81, 55, 29, 63, 40])

selected = ages[[2, 4, 7, 10]]
mask = ages >= 60
senior = ages[mask]
filtered = ages[(ages >= 20) & (ages <= 50)]

print("Selected Ages:", selected)
print("Senior Patients:", senior)
print("Number of Senior Patients:", np.sum(mask))
print("Age 20 to 50:", filtered)
print("Average Age:", np.mean(filtered))
```

```python
import numpy as np

sales = np.array([45000, 52000, 38000, 67000, 72000, 49000,
                  81000, 56000, 43000, 90000, 61000, 75000])

selected = sales[[1, 4, 7, 10]]
mask = sales > 60000
filtered = sales[(sales >= 50000) & (sales <= 75000)]

print("Selected Sales:", selected)
print("Sales above ₹60,000:", sales[mask])
print("Number of Months:", np.sum(mask))
print("Filtered Sales:", filtered)
print("Average Sales:", np.mean(filtered))
print("Highest Filtered Sale:", np.max(filtered))
```
