```python
import pandas as pd
import numpy as np

```

```python
data = {
    'Student_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Name': ['Arun', 'Priya', 'Karthik', 'Divya', 'Ravi', 'Sneha', 'Manoj', 'Anitha'],
    'Department': ['CSE', 'CSE', 'DS', 'DS', 'AI', 'AI', 'CSE', 'DS'],
    'Attendance': [85, 90, np.nan, 88, np.nan, 92, 76, np.nan],
    'Internal_Marks': [24, 26, 22, np.nan, np.nan, 28, 20, 25],
    'Assignment_Marks': [18, np.nan, 17, 19, np.nan, 20, np.nan, 18],
    'Final_Marks': [78, 85, 72, 80, np.nan, 91, 65, 74]
}
```

```python
ds = pd.DataFrame(data)
```

```python
print("1. After dropna():")
print(ds.dropna())
```

```python
print("\n2. After dropna(how='all'):")
print(ds.dropna(how='all'))
```

```python
ds['Remarks'] = np.nan
print("\n3. After removing empty columns:")
print(ds.dropna(axis=1, how='all'))
```

```python

print("\n4. After fillna(0):")
print(ds.fillna(0))

```

```python
print("\n5. Missing values:")
print(ds.isnull().sum())

```

```python
print("\n6. Available values:")
print(ds.notnull().sum())

```

```python
print("\n7. Forward Fill Attendance:")
print(ds['Attendance'].ffill())

```

```python
print("\n8. Backward Fill Attendance:")
print(ds['Attendance'].bfill())
```
