```python
import pandas as pd

student = pd.DataFrame({
    'Student_ID':[101,102,103,104,105,106],
    'Name':['Arun','Priya','Karthik','Divya','Ravi','Sneha'],
    'Department':['CSE','CSE','Data Science','Data Science','AI','AI']
})

marks = pd.DataFrame({
    'Student_ID':[101,102,103,104,107],
    'Internal_Marks':[24,26,22,25,21],
    'Assignment_Marks':[18,20,17,19,16],
    'Final_Marks':[78,85,72,80,68]
})

print(pd.merge(student, marks, on='Student_ID', how='inner'))
print(pd.merge(student, marks, on='Student_ID', how='left'))
print(pd.merge(student, marks, on='Student_ID', how='right'))
print(pd.merge(student, marks, on='Student_ID', how='outer'))
```
