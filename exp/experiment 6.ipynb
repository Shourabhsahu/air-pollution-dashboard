```python

import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat']
visitors = [500, 650, 600, 750, 900, 1000]
page_views = [1200, 1500, 1400, 1800, 2200, 2500]

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(days, visitors, marker='o')
plt.title('Visitors per Day')
plt.xlabel('Days')
plt.ylabel('Visitors')
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(days, page_views, marker='s')
plt.title('Page Views per Day')
plt.xlabel('Days')
plt.ylabel('Page Views')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(days, visitors, marker='o', label='Visitors')
plt.plot(days, page_views, marker='s', label='Page Views')
plt.title('Visitors vs Page Views')
plt.xlabel('Days')
plt.ylabel('Count')
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 4)
plt.plot(days, visitors, marker='o', linestyle=':')
plt.title('Visitors Trend')
plt.xlabel('Days')
plt.ylabel('Visitors')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

quarters = ['Q1', 'Q2', 'Q3', 'Q4']

revenue = [120, 150, 180, 220]
profit = [30, 40, 50, 65]
customers = [1000, 1250, 1500, 1800]

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(quarters, revenue, marker='o')
plt.title('Quarterly Revenue')
plt.xlabel('Quarter')
plt.ylabel('Revenue (₹ Lakhs)')
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(quarters, profit, marker='s', linestyle='--')
plt.title('Quarterly Profit')
plt.xlabel('Quarter')
plt.ylabel('Profit (₹ Lakhs)')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(quarters, customers, marker='^')
plt.title('Quarterly Customers')
plt.xlabel('Quarter')
plt.ylabel('Customers')
plt.grid(True)

plt.subplot(2, 2, 4)
plt.plot(quarters, revenue, marker='o', label='Revenue')
plt.plot(quarters, profit, marker='s', label='Profit')
plt.title('Revenue vs Profit')
plt.xlabel('Quarter')
plt.ylabel('Amount (₹ Lakhs)')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
