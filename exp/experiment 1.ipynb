```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Vertical Stack:\n", np.vstack((a, b)))
print("Horizontal Stack:", np.hstack((a, b)))

x = np.array([10, 25, 35, 45, 55, 65])
print("Filtered Array:", x[(x > 20) & (x < 60)])

y = np.array([0, 1, 2, 1, 0])
print("One-Hot Encoding:\n", np.eye(3, dtype=int)[y])
```
