```python
import numpy as np

salary = np.array([30000, 35000, 40000, 45000, 50000])

bonus = np.array([0.10, 0.15, 0.20])

bonus_amount = salary[:, np.newaxis] * bonus
final_salary = salary[:, np.newaxis] + bonus_amount

print("Bonus Matrix:")
print(bonus_amount)

print("\nFinal Salary Matrix:")
print(final_salary)
```
